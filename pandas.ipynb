```python
import pandas as pd
print("Pandas version: ", pd.__version__)
```

```python
# first example
mydataset = {
    'cars': ["VOlvo", "BMW", "Audi"],
    'passing': [3,5,7]
}

myvar = pd.DataFrame(mydataset)
print(myvar)
```

```python
# Pandas Series --> like a column in a table
# It's a one-dimensional array holding data of any type

s = ["a string", 2, 5, 6.9]
myvar = pd.Series(s)
print(myvar)
```

```python
# Creates Labels: with the index argument, we can name out own labels
s = ["a string", 2, 6.9]
myvar = pd.Series(s, index = ["first", "second", "third"])
print(myvar)
print("The second data is: ", myvar["second"])
```

```python
# Key/Value object as a series
# You can also use a key/value object, like a dictionary, when creating a Series.

calories = {
    'day 1': 452,
    'day 2': 343,
    'day 3': 234
}

myvar = pd.Series(calories)
print(myvar)

# must use the same name as the keys
myvar2 = pd.Series(calories, index = ["day 1", "day 2"])
print(myvar2)
```

```python
# Data sets in Pandas are usually multi-dimensional tables, called DataFrames.
# Series is like a column, a DataFrame is the whole table.

data = {
    'calories': [234234,56546,6474676],
    'duration': [3,4,5]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Pandas use the loc attribute to return one or more specified row(s)
print(df.loc[0], "\n")
print(df.loc[1], "\n")
print(df.loc[2], "\n")
print(df.loc[[0,1]], "\n")
print(df*10)
```
