```python
# how to save 2 matrices in a file?
# always use binary format, not text format!!!

'''
Text format
    np.savetxt

Binary Format
    np.save --> .npy standard format
    np.savez --> .npz uncompressed format (for many arrays)
    np.savez_compressed --> compressed version of savez
    np.load --> loading .npy or .npz
'''

X = np.random.randint(1, 100, size=(5,5))
Y = np.random.randint(1, 100, size=(5,5))

np.savez('data.npz', X=X, Y=Y) # saves in binary format
data = np.load('data.npz')

np.savetxt('readable.txt', X, fmt="%d") # saves in human text

print(data['X'])
print(data['Y'])
```
